```python
url = "https://api.nasa.gov/neo/rest/v1/feed"

params = {
    "start_date": "2026-09-01",
    "end_date": "2026-09-07",
    "api_key": API_KEY
}

response = requests.get(url, params=params)

print("Status code:", response.status_code)
asteroid_data = response.json()

asteroid_data.keys()
```

```python
asteroid_data["near_earth_objects"].keys()
```

```python
first_date = list(asteroid_data["near_earth_objects"].keys())[0]

asteroid_data["near_earth_objects"][first_date]
```

```python
asteroids = []
```

```python
for date, asteroid_list in asteroid_data["near_earth_objects"].items():

    for asteroid in asteroid_list:

        asteroid_id = asteroid["id"]
        asteroid_name = asteroid["name"]
        absolute_magnitude = asteroid["absolute_magnitude_h"]

        min_diameter = asteroid["estimated_diameter"]["kilometers"]["estimated_diameter_min"]

        relative_velocity = asteroid["close_approach_data"][0]["relative_velocity"]["kilometers_per_second"]

        asteroids.append({
            "Asteroid ID": asteroid_id,
            "Asteroid name": asteroid_name,
            "The Minimal estimated diameter in Kilometre": min_diameter,
            "Absolute_magnitude": absolute_magnitude,
            "Relative_velocity(km/s)": relative_velocity
        })
```

```python
df = pd.DataFrame(asteroids)

df
```

```python
df.head()
df.shape
df.columns
df.info()
print(df.columns.tolist())
```

```python
numeric_columns = [
    "The Minimal estimated diameter in Kilometre",
    "Absolute_magnitude",
    "Relative_velocity(km/s)"
]

df[numeric_columns] = df[numeric_columns].apply(pd.to_numeric, errors="coerce")
```

```python
df = df.reset_index(drop=True)
df.head()
```
